```python
model = YOLO("yolo11l.pt")
model.train(
    data="/kaggle/working/car-plates-detection-1/data.yaml",
    epochs=80,
    imgsz=256,
    batch=16,
    amp = True,
    cls= 1.5,
    label_smoothing=0.1

)
```
